# Section 1 — Talking to an LLM Programmatically

Before building anything, understand the anatomy of an API call: **messages and roles**
(`system`, `user`, `assistant`), and the **generation parameters** (`temperature`,
`max_tokens`).

### Part 1.1 — Your first API call

````python
# TODO: Write a helper function you will reuse for the WHOLE lab:
def ask_llm(user_prompt, system_prompt="You are a helpful assistant.",
            temperature=0.7, max_tokens=500):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": user_prompt},
        ],
        temperature=temperature,
        max_tokens=max_tokens,
    )

    # TODO: Print response.usage as well — how many tokens did your call consume?
    print(f"Token Usage -> Prompt: {response.usage.prompt_tokens} | Completion: {response.usage.completion_tokens} | Total: {response.usage.total_tokens}")

    return response.choices[0].message.content

# TODO: Call it once with a simple question and print the answer.
print(ask_llm("What is GitHub?"))
````

**Student Reasoning — Anatomy of a call**
*1. What is the difference between the `system` and `user` roles? Give an example of
something that belongs in each.*
*2. What is a token, roughly? Why do API providers bill per token rather than per request?*

> **Answer:**
1. The system role sets the overarching behavior, constraints, and persona of the AI like "You are a concise financial analyst.". The user role provides the specific input or task for that instance,for instance, "Summarize this loan application: [text]".
2. A token represents a chunk of text, roughly equivalent to a word or part of a word. API providers bill per token because their computational costs directly scale with the volume of text processed (both input context and output generation), rather than the arbitrary length of a single request.

### Part 3.2 — Component 2: Structured extraction (JSON)
Downstream software cannot read prose. Extract the fields in `GOLD` as strict JSON.

````python
# TODO: Write EXTRACT_PROMPT — a template that instructs the model to return ONLY a JSON
#   object with EXACTLY these keys:
#     applicant_name (string), amount_ghs (number), purpose (string),
#     monthly_profit_ghs (number or null), has_collateral_or_guarantor (boolean),
#     repayment_months (number or null)
#   Techniques to use:
#     - explicit schema in the prompt
#     - ONE worked example (few-shot) using a letter you write yourself (not from LETTERS!)
#     - "If a field is not stated in the letter, use null. Do not guess."
#     - temperature=0

import json
import pandas as pd

EXTRACT_PROMPT_SYS = """You are a data extraction bot for a microfinance institution.
Extract the following fields from the loan application letter into strict JSON format with exactly these keys:
- "applicant_name" (string)
- "amount_ghs" (number)
- "purpose" (string)
- "monthly_profit_ghs" (number or null)
- "has_collateral_or_guarantor" (boolean)
- "repayment_months" (number or null)

If a field is not stated in the letter, use null. Do not guess. Output ONLY raw JSON.

Example Input:
My name is Gerard. I need GHS 5000 for farming tools. I don't have collateral. I make GHS 150 profit a month.

Example Output:
{
  "applicant_name": "Gerard",
  "amount_ghs": 5000,
  "purpose": "farming tools",
  "monthly_profit_ghs": 150,
  "has_collateral_or_guarantor": false,
  "repayment_months": null
}"""

EXTRACT_PROMPT_USER = "{letter_text}" # Format to pass in letters

# TODO: Write extract_fields(letter_text) that calls the LLM, strips any ```json fences,
#   json.loads() the result, and returns a dict. Handle parse failures gracefully
#   (return None and print a warning).
def extract_fields(letter_text):
    raw_response = ask_llm(
        user_prompt=EXTRACT_PROMPT_USER.format(letter_text=letter_text),
        system_prompt=EXTRACT_PROMPT_SYS,
        temperature=0.0
    )
    clean_json = raw_response.replace("```json", "").replace("```", "").strip()
    try:
        return json.loads(clean_json)
    except json.JSONDecodeError:
        print("Warning: Failed to parse JSON")
        return None

# TODO: Run it on ALL SIX letters; collect results into a pandas DataFrame (one row per
#   letter) and display it.
results = [extract_fields(letter) for letter in LETTERS.values()]
df = pd.DataFrame([r for r in results if r])
print(df)
````

**Student Reasoning — Structured extraction**
*1. Why must the few-shot example NOT come from the six letters you are processing?*
*2. Why "use null, do not guess" — what did the model do without that instruction?*
*3. Why is temperature=0 the right choice for extraction but arguably not for creative tasks?*

> **Answer:**
1. Using one of the actual dataset letters as the few-shot example introduces data leakage. If the model is prompted to extract data from a letter it already has the "answers" for in its system prompt, it might overfit to that specific example. It could also accidentally pull details from the few-shot example and insert them into the extraction for a different letter.
2. Without this instruction, LLMs inherently try to be "helpful" by inferring or fabricating missing data to complete the schema. For example, looking at your DataFrame, both Kwame Boateng and Kofi correctly returned NaN (parsed from null) for monthly_profit_ghs, and Kwame returned NaN for repayment_months because they didn't state these figures. If the model were allowed to guess, it might have hallucinated a standard 12-month repayment plan for Kwame, which would corrupt the financial data with false information.
3. Temperature=0 forces the model to be entirely deterministic, choosing the most probable tokens every time. This is critical for data extraction because downstream code (like json.loads()) requires strict, predictable syntax. As seen in your output, a temperature of 0 successfully produced perfectly formatted JSON for all six distinct letters, allowing them to cleanly parse into a Pandas DataFrame without any decoding errors. For creative tasks like storytelling or brainstorming, temperature=0 would result in flat, repetitive, and boring text, where higher temperatures are needed to introduce varied vocabulary and ideas.

### Part 3.3 — Component 3: The decision-support brief
Combine everything: for each letter, produce a recommendation brief for the loan officer —
strengths, risks, missing information, and a suggested next step. The system must
**support** the decision, not **make** it.

````python
# TODO: Write BRIEF_PROMPT — it receives the letter AND your extracted JSON, and must output:
#     1. Strengths (bullet points, grounded in the letter)
#     2. Risks / red flags (bullet points)
#     3. Missing information the officer should request
#     4. Suggested next step (e.g. "invite for interview", "request documents",
#        "flag for senior review") — NOT "approve" or "reject".
#   Give the model an explicit instruction that final decisions are made by humans.

BRIEF_PROMPT_SYS = """You are a decision-support assistant for a microfinance loan officer.
Your job is to analyze a loan application letter and its extracted data to provide a neutral, objective briefing.

CRITICAL INSTRUCTION: Final decisions are made by human officers. You MUST NOT output "approve", "reject", or suggest a final approval status.

Format your response EXACTLY with these four headings:
- Strengths: (Bullet points, strictly grounded in the text)
- Risks: (Bullet points highlighting red flags or weaknesses)
- Missing Information: (Specific details the officer needs to request)
- Suggested Next Step: (A single action-oriented step, e.g., "invite for interview", "request documents", "flag for senior review")"""

BRIEF_PROMPT_USER = """Applicant Letter:
{letter_text}

Extracted Data:
{json_data}"""

# TODO: Generate briefs for ALL SIX letters. Print the briefs for L001, L002, and L006 —
#   three very different applications.

print("Generating briefs...\n" + "="*50)
briefs = {}

for (letter_id, letter_text), (_, row) in zip(LETTERS.items(), df.iterrows()):
    # Convert the row back to a JSON string for the prompt
    json_data = row.to_json()

    formatted_user_prompt = BRIEF_PROMPT_USER.format(
        letter_text=letter_text,
        json_data=json_data
    )

    # Call the LLM at temperature 0 for consistent formatting
    brief = ask_llm(
        user_prompt=formatted_user_prompt,
        system_prompt=BRIEF_PROMPT_SYS,
        temperature=0.0
    )
    briefs[letter_id] = brief

# Print the requested briefs
for letter_id in ["L001", "L002", "L006"]:
    print(f"\n=== Brief for {letter_id} ===")
    print(briefs[letter_id])
````
